```python
import numpy as np
import pandas as pd
import statsmodels.formula.api as smf

# Reproducibility
np.random.seed(0)

# Number of observations
n = 50

# Generate p and q
p = np.random.randint(0, 20, size=n)
q = np.random.randint(0, 20, size=n)

# Create DataFrame
df = pd.DataFrame({
    "p": p,
    "q": q
})

# Regression
model = smf.ols("p ~ q", data=df).fit()

# Output
print(model.summary())

# Interpretation of the regression results for p ~ q using the df DataFrame:
# R-squared (0.014): This indicates that 'q' explains a very small proportion (1.4%) of the variance in 'p'.
# This suggests that 'q' is not a strong linear predictor of 'p' in this model.

# Coefficient for 'q' (0.1406): For every one-unit increase in 'q', 'p' is estimated to increase by 0.1406 units.
# However, the P-value for 'q' (0.407) is high (greater than 0.05), which means this coefficient is not statistically significant.

# Intercept (8.0424): When 'q' is 0, the expected value of 'p' is 8.0424. This intercept is statistically significant (P-value < 0.001).
```
